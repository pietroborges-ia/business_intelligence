```python
import numpy as np
```

```python
A = np.array(((2,2,1), (1,3,1), (4, 6, 1)))
print(A)
```

```python
det = np.linalg.det(A)
print(det)
```

```python
det = np.linalg.det(A)
if det == 0:
    print(f"Os pontos são colineares, pois o det = {det}")
else:
    print(f"Os pontos não são colineares, pois o det = {det}")
```
