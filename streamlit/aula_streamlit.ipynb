```python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("agro.csv", encoding = 'utf-8', sep = ";")
df.head()
```

```python
df = df.replace('-', '0')
df.head()
```

```python
df.dtypes
```

```python
# df.iloc[:, 1:] seleciona todas as colunas, exceto a primeira
# col.str.replace('.', '', regex=False) remove os pontos e
# col.str.replace(',', '.', regex=False) troca a vírgula por ponto
cols = df.columns[2:]  # Seleciona todas as colunas, exceto a primeira
df[cols] = df[cols].apply(
    lambda col: pd.to_numeric(
        col.str.replace('.', '', regex=False).str.replace(',', '.', regex=False),
        errors='coerce'
    )
)

df.head(10)
```

```python
tipos = df['2020'].apply(type)
tipos.head(10)
```

```python
numeric_cols = df.select_dtypes(include=['float', 'int']).columns
df[numeric_cols] = df[numeric_cols].round(2)
df.head()
```
