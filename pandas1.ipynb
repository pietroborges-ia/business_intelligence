```python
import pandas as pd
import numpy as np
```

#### Questão 1

```python
# Dicionário contendo os nomes e idades de cinco alunos
dados_escola = {
    "Nome": ["Pietro", "Marjory", "Maressa", "Gege", "Jene"],
    "Idade": [18, 19, 20, 21, 22]
}

# Criando o DataFrame a partir do dicionário
df_alunos = pd.DataFrame(dados_escola)

# Exibindo o DataFrame final com o índice automático (0, 1, 2, 3, 4)
print(df_alunos)
```

#### Questão 2

```python
# Lista de dicionários contendo as informações dos funcionários
dados_rh = [
    {'Nome': 'Ana', 'Cargo': 'Analista'},
    {'Nome': 'Bruno', 'Cargo': 'Gerente', 'Bonus': 500}
]

# Transformando a lista em um DataFrame
df_rh = pd.DataFrame(dados_rh)

# Exibindo o DataFrame final
print(df_rh)

# EXPLICAÇÃO:
# O que acontece com a coluna 'Bonus' para a funcionária Ana?
# Como o dicionário correspondente à Ana não possui a chave 'Bonus', 
# o Pandas preenche automaticamente essa lacuna com 'NaN' (Not a Number), 
# que é a forma padrão da biblioteca representar valores nulos ou ausentes.
```

#### Questão 3

```python
# Criando a primeira Series com os nomes dos produtos
nomes_produtos = pd.Series(["Notebook", "Smartphone", "Tablet"], index=["p1", "p2", "p3"])

# Criando a segunda Series com os preços
precos_produtos = pd.Series([4500.00, 2500.00, 1500.00], index=["p1", "p2", "p3"])

# Unindo as Series em um DataFrame
df_produtos = pd.DataFrame({
    "Produto": nomes_produtos,
    "Preço": precos_produtos
})

print(df_produtos)
```

#### Questão 4

```python
# Criando um DataFrame de exemplo para demonstrar o código
dados = {
    "Sede_A": [1500.0, 3200.0, 4100.0],
    "Sede_B": [2000.0, 2800.0, 3900.0]
}
df = pd.DataFrame(dados)

# 1. Criando a coluna "Total_Vendas" com a soma das duas sedes
df["Total_Vendas"] = df["Sede_A"] + df["Sede_B"]

# 2. Adicionando a coluna "Imposto" calculada como 10% do valor da "Sede_A"
df["Imposto"] = df["Sede_A"] * 0.10

print(df)
```

#### Questão 5

```python
# Criando um DataFrame de exemplo para a demonstração
dados_clientes = {
    "Nome": ["Pietro", "Marjory", "Maressa"],
    "CPF": ["111.111.111-11", "222.222.222-22", "333.333.333-33"],
    "Telefone": ["9999-1111", "9888-2222", "9777-3333"]
}
df = pd.DataFrame(dados_clientes)

# 1. Excluindo a coluna "CPF" usando o comando del
del df["CPF"]

# 2. Excluindo a coluna "Telefone" usando o método pop()
telefones_removidos = df.pop("Telefone")

# Exibindo o conteúdo da coluna removida
print(telefones_removidos)
```

#### Questão 6

```python
# Criando um DataFrame de exemplo onde o índice é o nome do país
dados_paises = {
    "Populacao": [214000000, 45000000, 19000000],
    "Capital": ["Brasília", "Buenos Aires", "Santiago"]
}
df_paises = pd.DataFrame(dados_paises, index=["Brasil", "Argentina", "Chile"])

# Utilizando o método loc para recuperar apenas os dados do "Brasil"
dados_brasil = df_paises.loc["Brasil"]

# Imprimindo os dados recuperados
print(dados_brasil)
```

#### Questão 7

```python
# Criando um DataFrame com dados de 6 modelos de carros
dados_carros = {
    "Modelo": ["Fusca", "Gol", "Palio", "Onix", "HB20", "Corolla"],
    "Marca": ["Volkswagen", "Volkswagen", "Fiat", "Chevrolet", "Hyundai", "Toyota"]
}
df_carros = pd.DataFrame(dados_carros)

# Utilizando o indexador iloc para selecionar o quarto (índice 3) e o quinto (índice 4) carro
# O fatiamento 3:5 inclui o índice 3 e vai até o 4 (exclusivo no 5)
carros_selecionados = df_carros.iloc[3:5]

print(carros_selecionados)
```

#### Questão 8

```python
# Criando um DataFrame de exemplo com 20 linhas
dados_historicos = {
    "ID": range(1, 21),
    "Valor": range(100, 2100, 100)
}
df_original = pd.DataFrame(dados_historicos)

# Utilizando o operador de slicing para extrair da 5ª até a 12ª linha
# Lembre-se: a 5ª linha corresponde ao índice 4, e o fatiamento até 12 é exclusivo no limite superior (pega até o índice 11, que é a 12ª linha)
df_fatiado = df_original[4:12]

# Exibindo o novo DataFrame
print(df_fatiado)
```

#### Questão 9

```python
# Criando DataFrames de exemplo para loja 1 e loja 2
dados_loja1 = {
    "Produto": ["Camiseta", "Calça", "Tênis"],
    "Estoque": [50, 30, 20]
}
df_loja1 = pd.DataFrame(dados_loja1)

dados_loja2 = {
    "Produto": ["Meia", "Jaqueta", "Boné"],
    "Estoque": [100, 15, 40]
}
df_loja2 = pd.DataFrame(dados_loja2)

# Concatenando (anexando) os dados da loja2 ao final da loja1
# Utiliza-se pd.concat em vez do antigo .append() (que foi descontinuado nas versões recentes do Pandas)
# O parâmetro ignore_index=True recria os índices para ficarem sequenciais no novo DataFrame
df_concatenado = pd.concat([df_loja1, df_loja2], ignore_index=True)

# Exibindo o resultado
print(df_concatenado)
```

#### Questão 10

```python
# Criando o DataFrame com nomes de funcionários e salários
dados_funcionarios = {
    "Nome": ["Alice", "Bob", "Carlos", "Diana", "Eva"],
    "Salario": [4500.00, 5200.00, 3800.00, 6100.00, 4900.00]
}
df = pd.DataFrame(dados_funcionarios)

# 1. Imprimindo a transposta do DataFrame
print("1. Transposta do DataFrame:")
print(df.T)
print("\n")

# 2. Imprimindo os tipos de dados (dtypes) de cada coluna
print("2. Tipos de dados (dtypes):")
print(df.dtypes)
print("\n")

# 3. Imprimindo a quantidade total de elementos presentes no objeto (size)
print("3. Quantidade total de elementos (size):")
print(df.size)
print("\n")

# 4. Imprimindo apenas as duas últimas linhas do DataFrame (tail)
print("4. Duas últimas linhas (tail):")
print(df.tail(2))
```
