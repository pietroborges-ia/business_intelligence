# Numpy - Array (VETORES)

```python
import numpy as np
```

```python
#funções universais, aplicadas a todos os elementos

#Mostra o valor da raiz quadrada de todos elementos
k1 = np.array([1, 4, 9, 16, 25, 36])
print(np.sqrt(k1))
```

```python
#extração de elementos
m = np.array([11, 22, 33, 44, 55, 66])
print(m)

#Mostra o array criado a partir da posição 0, dois elementos
print(m[0:2])
```

```python
# Mostra o array criado a partir da 2° posição
#até todo o restante do array
m = np.array([1, 2, 3, 4, 5, 6])
print(m)
print(m[1:])
```

```python
# Mostra o array criado a partir da antepenúltima
#posiçã até o final
m = np.array([1, 2, 3, 4, 5, 6])
print(m)
print(m[-3:])
```

```python
#extração de linhas e colunas
l = np.array([[4, 5], [6, 1], [7, 4]])
print(l)
```

```python
#primeira linha, todas as colunas
l_linha_1 = l[0, :]
print(l)
print("Essa é a primeira linha", l_linha_1)
```

```python
#segunda linha
l_linha_2 = l[1, :]
print(l)
print("Essa é a segunda linha:", l_linha_2)
```

```python
#terceira linha
l_linha_3 = l[2, :]
print(l)
print("Essa é a terceira linha:", l_linha_3)
```

```python
#todas as linhas, primeira coluna
l_coluna_1 = l[:, 0]
print("Elementos da primeira coluna", l_coluna_1)
```

```python
#todas as linhas, segunda coluna
l_coluna_2 = l[:, 1]
print("Elementos da segunda coluna", l_coluna_2)
```
