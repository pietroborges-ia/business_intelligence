#### Exercício 5 - Crie uma função lambda e atribua seu retorno a uma variável chamada soma. A expressão irá receber 2 números como parâmetros e retornar a soma deles.

```python
soma = lambda a, b: a + b

num1 = int(input("Digite o primeiro número: "))
num2 = int(input("Digite o segundo número: "))

print("A soma é:", soma(num1, num2))
```

### Exercício 6. Lista com temperaturas em graus Celsius. 
### Crie uma função lambda que converta cada temperatura para Fahrenheit. 

Em física do ensino médio, é possível conseguir a fórmula matemática que converte, a temperatura, de Celsius para Fahrenheit?

R: Sim. A fórmula de conversão ensinada no ensino médio é: F=(C×5|9)+32

```python
def converter_temperaturas():
    Celsius = [39.2, 36.5, 37.3, 37.8]

    converter = lambda c: (c * 9/5) + 32

    Fahrenheit = list(map(converter, Celsius))

    print(Fahrenheit)

converter_temperaturas()
```

### Exercício 7.

#### Faça um programa que tenha uma função chamada amplitude. A função deve receber uma lista e imprimir a amplitude.

```python
def amplitude(lista):
    amp = max(lista) - min(lista)
    print("A amplitude é:", amp)

lista = []

quantidade = int(input("Quantos números deseja adicionar à lista? "))

for i in range(quantidade):
    numero = int(input(f"Digite o {i + 1}º número: "))
    lista.append(numero)

amplitude(lista)
```

### Exercício 8 - Construir uma função que imprima uma sequência.

```python
def sequencia(n):
    for i in range(1, n + 1):
        for j in range(i):
            print(i, end=" ")
        print()

n = int(input("Digite um número inteiro: "))
sequencia(n)
```
