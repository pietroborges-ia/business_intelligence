## Atividade 2

```python
# --- ATIVIDADE 2: COLETA E REGRAS DE PROMOÇÃO DE ESTOQUE ---

# Passo 1 do BI: Estruturação dos dados coletados sobre o estoque
produtos = [
    {
        "id": 101,
        "nome": "Liquidificador Turbo",
        "estoque": 85,
        "dias_no_estoque": 120,
        "nova_versao_chegando": True,
        "custo": 120.0,
        "preco_venda": 200.0
    },
    {
        "id": 102,
        "nome": "Sanduicheira Grill",
        "estoque": 20,
        "dias_no_estoque": 15,
        "nova_versao_chegando": False,
        "custo": 60.0,
        "preco_venda": 100.0
    },
    {
        "id": 103,
        "nome": "Air Fryer Digital",
        "estoque": 110,
        "dias_no_estoque": 95,
        "nova_versao_chegando": False,
        "custo": 250.0,
        "preco_venda": 450.0
    },
    {
        "id": 104,
        "nome": "Batedeira Planetária",
        "estoque": 15,
        "dias_no_estoque": 40,
        "nova_versao_chegando": False,
        "custo": 180.0,
        "preco_venda": 320.0
    },
    {
        "id": 105,
        "nome": "Cafeteira Expressa",
        "estoque": 90,
        "dias_no_estoque": 110,
        "nova_versao_chegando": True,
        "custo": 150.0,
        "preco_venda": 280.0
    }
]

print("=== PRODUTOS ELEGÍVEIS PARA PROMOÇÃO (REGRA DE BI) ===")

# Regra de Negócio:
# Produto entra em oferta se tiver estoque > 80
# OU se uma nova versão estiver chegando.
for p in produtos:
    motivo = []

    if p["estoque"] > 80:
        motivo.append(f"Estoque alto ({p['estoque']} un)")

    if p["nova_versao_chegando"]:
        motivo.append("Nova versão em lançamento")

    if p["dias_no_estoque"] > 90 and p["estoque"] > 50:
        motivo.append("Desconto Agressivo de 25%")

    if "Desconto Agressivo de 25%" in motivo:
        preco_com_desconto = p["preco_venda"] * 0.75

    if motivo:
        print(f"-> Produto: {p['nome']} (ID: {p['id']})")
        print(f"   Motivos para Promoção: {', '.join(motivo)}")
        print(f"   Preço Atual: R$ {p['preco_venda']:.2f}")

        if "Desconto Agressivo de 25%" in motivo:
            print(f"   Preço com 25% de desconto: R$ {preco_com_desconto:.2f}")

        print()
```

**Por que a estruturação de dados em formatos como dicionários e tabelas é
indispensável para o trabalho de um especialista em BI e IA?**

A estruturação dos dados em formatos como dicionários e tabelas é essencial porque facilita a organização e o acesso às informações. Quando os dados estão bem estruturados, fica mais fácil fazer consultas, aplicar regras de negócio, gerar relatórios e encontrar padrões.

No Business Intelligence (BI), isso ajuda as empresas a analisarem seus dados e tomarem decisões com mais rapidez e segurança. Já na Inteligência Artificial (IA), os dados organizados são fundamentais para treinar modelos, fazer previsões e obter resultados mais precisos. Sem uma boa estruturação, as análises ficam mais difíceis, sujeitas a erros e menos confiáveis.

## Atividade 3

```python
import pandas as pd

# --- ATIVIDADE 3: ANÁLISE DESCRITIVA DE VENDAS COM PANDAS ---

# Dados de vendas diárias da loja (Passo 1: Coleta)
dados_vendas = {
    "Dia_Semana": [
        "Segunda",
        "Terça",
        "Quarta",
        "Quinta",
        "Sexta",
        "Sábado",
        "Domingo"
    ],
    "Faturamento_R$": [
        12500.00,
        14200.00,
        11800.00,
        15600.00,
        22400.00,
        28900.00,
        9500.00
    ],
    "Clientes_Atendidos": [
        110,
        125,
        105,
        130,
        180,
        220,
        80
    ],
    "Ticket_Medio_R$": [0, 0, 0, 0, 0, 0, 0]
}

# Criando o DataFrame (Passo 2: Preparo e Análise)
df_vendas = pd.DataFrame(dados_vendas)

# Cálculo da métrica derivada: Ticket Médio (Faturamento / Clientes)
df_vendas["Ticket_Medio_R$"] = (
    df_vendas["Faturamento_R$"] / df_vendas["Clientes_Atendidos"]
)

# Extração de estatísticas descritivas do BI
faturamento_total = df_vendas["Faturamento_R$"].sum()
faturamento_medio = df_vendas["Faturamento_R$"].mean()
dia_campeao = df_vendas.loc[df_vendas["Faturamento_R$"].idxmax()]
mediana_faturamento = df_vendas["Faturamento_R$"].median()

print("=" * 55)
print(" RELATÓRIO DE DESEMPENHO SEMANAL DE BI")
print("=" * 55)
print(df_vendas.to_string(index=False))
print("-" * 55)
print(f"Faturamento Total da Semana: R$ {faturamento_total:,.2f}")
print(f"Faturamento Médio Diário: R$ {faturamento_medio:,.2f}")
print(f"Mediana do Faturamento: R$ {mediana_faturamento:,.2f}")
print(f"Dia de Pico de Vendas: {dia_campeao['Dia_Semana']} (R$ {dia_campeao['Faturamento_R$']:,.2f})")
print("=" * 55)
```

**Compare a média diária (R$ 16.414,29) com a mediana. Por que existe uma diferença entre elas e qual insight de BI essa diferença revela sobre o comportamento de compras no fim de semana?**

A média do faturamento foi de R$ 16.414,29, um valor maior que a mediana. Isso acontece porque as vendas de sexta-feira e sábado foram significativamente mais altas, elevando a média.

Insight de BI: Para planejar a escala diária de funcionários, o gestor não deve considerar apenas a média, pois os resultados dos fins de semana aumentam esse valor. Nesse caso, a mediana representa melhor o faturamento de um dia útil comum e pode servir como uma referência mais adequada para o planejamento.


## Atividade 4

```python
# --- ATIVIDADE 4: CÁLCULO E AUTOMAÇÃO DE ROI EM BI ---

def analisar_projeto_bi(nome_projeto, custo_investimento, receita_gerada):
    """
    Função que calcula o Lucro Líquido e o ROI de um projeto de BI
    e emite uma recomendação estratégica automatizada.
    """

    lucro_liquido = receita_gerada - custo_investimento
    roi_percentual = (lucro_liquido / custo_investimento) * 100

    print(f"PROJETO: {nome_projeto}")
    print(f" -> Custo do Investimento: R$ {custo_investimento:,.2f}")
    print(f" -> Receita Gerada: R$ {receita_gerada:,.2f}")
    print(f" -> Lucro Líquido: R$ {lucro_liquido:,.2f}")
    print(f" -> ROI Calculado: {roi_percentual:.2f}%")

    # Tomada de decisão automatizada com base em limiares de ROI
    if roi_percentual >= 100:
        decisao = "APROVADO COM EXCELÊNCIA (Projeto de Altíssimo Retorno)"
    elif roi_percentual > 0:
        decisao = "APROVADO (Projeto Lucrativo)"
    elif roi_percentual == 0:
        decisao = "PONTO DE EQUILÍBRIO (Sem Lucro nem Prejuízo)"
    else:
        decisao = "REPROVADO (Projeto Gerou Prejuízo)"

    print(f" -> Decisão Estratégica: {decisao}")
    print("-" * 55)

    return roi_percentual


# Avaliando 3 cenários de campanhas da loja de eletrodomésticos
print("=== AVALIAÇÃO DE PROJETOS E CAMPANHAS DE BI ===")

analisar_projeto_bi(
    "Campanha A: Anúncios em Redes Sociais (Air Fryer)",
    custo_investimento=5000.00,
    receita_gerada=18000.00
)

analisar_projeto_bi(
    "Campanha B: Implementação de Dashboard de Estoque",
    custo_investimento=12000.00,
    receita_gerada=28000.00
)

analisar_projeto_bi(
    "Campanha C: Promoção sem Análise de Dados (Desconto Geral)",
    custo_investimento=8000.00,
    receita_gerada=6000.00
)
```

**Calcule manualmente o ROI do Projeto B: Custo = 12.000,00 reais | Receita = 28.000,00 reais. O resultado do código confere com a sua conta manual?**

O lucro líquido do Projeto B é de R$ 16.000,00. Aplicando a fórmula do ROI, temos:

ROI = (16.000 ÷ 12.000) × 100 = 133,33%

**Se o gestor da empresa tivesse apenas R$ 15.000,00 de orçamento disponível, qual
combinação de projetos traria o maior Lucro Líquido total? Justifique com base em
dados.**

Projeto A: Lucro = 13.000 reais | ROI = 260.00%
Projeto B: Lucro = 16.000 reais | ROI = 133.33%
Projeto C: Prejuízo = -2.000 reais | ROI = -25.00%
Resposta do Desafio do Orçamento (15.000 reais): O gestor pode executar a
Campanha A (Custo 5.000 reais) e a Campanha B (Custo 12.000 reais)? Não, pois a
soma seria 17.000 reais (acima de 15.000 reais). A melhor decisão com 15.000 reais de
orçamento disponível é aprovar o Projeto B sozinho (Lucro de 16.000 reais) ou o
Projeto A sozinho (Lucro 13.000 reais, sobrando 10.000 reais em caixa).
Decisão de BI: Nunca escolha apenas pelo ROI % absoluto sem checar o
orçamento total e a escala do Lucro Líquido gerado em Reais.

## Atividade 5

```python
import pandas as pd

# --- ATIVIDADE 5: PIPELINE COMPLETA DE BI E PROJEÇÃO PREDITIVA ---

# PASSO 1: COLETA DE DADOS (Dataset Histórico da Aula 1 - Statista/IDC)
dados_datasphere = {
    "Ano": [2020, 2021, 2022, 2023, 2024, 2025],
    "Volume_ZB": [64.2, 79.0, 97.0, 120.0, 147.0, 181.0]
}

df_bi = pd.DataFrame(dados_datasphere)

# PASSO 2: PREPARO E ANÁLISE DE DADOS (Cálculo da taxa de crescimento YoY)
df_bi["Crescimento_Absoluto_ZB"] = df_bi["Volume_ZB"].diff()
df_bi["Taxa_Crescimento_%"] = df_bi["Volume_ZB"].pct_change() * 100

# Média anual de crescimento percentual
taxa_media_crescimento = df_bi["Taxa_Crescimento_%"].mean()

# SINERGIA COM DATA SCIENCE: Projeção Preditiva Simples para 2026
volume_2025 = df_bi.loc[df_bi["Ano"] == 2025, "Volume_ZB"].values[0]
projecao_2026_zb = volume_2025 * (1 + (taxa_media_crescimento / 100))

# PASSO 3: VISUALIZAÇÃO DE DADOS (Dashboard em Texto com Gráfico de Barras ASCII)
print("=" * 65)
print(" DASHBOARD DE EVOLUÇÃO E PROJEÇÃO DO DATASPHERE GLOBAL (ZB)")
print("=" * 65)

for idx, row in df_bi.iterrows():
    ano = int(row["Ano"])
    vol = row["Volume_ZB"]
    barra = "█" * int(vol / 5)  # Cada bloco representa 5 Zettabytes

    if pd.notnull(row["Taxa_Crescimento_%"]):
        pct_str = f"+{row['Taxa_Crescimento_%']:.1f}%"
    else:
        pct_str = " N/A "

    print(f"{ano} | {vol:6.1f} ZB | {pct_str:7s} | {barra}")

# Linha da Projeção Preditiva
barra_proj = "█" * int(projecao_2026_zb / 5)
print("-" * 65)
print(f"2026 | {projecao_2026_zb:6.1f} ZB | Projeção | {barra_proj} (PREDITIVO)")
print("=" * 65)
```

**Executando a pipeline, qual foi a taxa média anual de crescimento de dados
calculada entre 2020 e 2025?**

A taxa média anual de crescimento de aproximadamente 23,03% ao ano** entre 2020 e 2025. Isso mostra que o volume de dados vem aumentando de forma consistente, com um crescimento médio de cerca de 23% a cada ano.

**Qual o valor estimado para o volume global de dados em 2026 de acordo com a
projeção preditiva?**

De acordo com a projeção preditiva da pipeline, o volume global de dados em 2026 será de aproximadamente 222,7 Zettabytes (ZB). Essa estimativa foi obtida aplicando a taxa média anual de crescimento de 23,03% sobre o volume registrado em 2025, que foi de 181 ZB.
