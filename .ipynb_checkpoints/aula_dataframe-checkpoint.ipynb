```python
import pandas as pd
```

```python
data = [1, 2, 3, 4, 5]
df = pd.DataFrame(data)
df
```

```python
data = [["Pietro", 25], ["Marjory", 26], ["Irineu", 13]]
df = pd.DataFrame(data, columns = ["Nome", "Idade"])
df
```

```python
data = [["Pietro", 25], ["Marjory", 26], ["Irineu", 13]]
df = pd.DataFrame(data, columns = ["Nome", "Idade"])
df["Idade"] = df["Idade"].astype(float)
df
```

```python
# Criando um DataFrame a partir de um Dicionário
data = {"Nome":["Pietro", "Peter", "Parker"], "Idade":[25, 28, 30]}
df = pd.DataFrame(data)
df
```

```python
# Usando função iloc do pandas
data = {"One" : pd.Series([1, 2, 3], index = ["a", "b", "c"]),
        "Two" : pd.Series([10, 20, 30, 40], index = ["a", "b", "c", "d"]),
        "Three" : pd.Series([1, 2, 3, 4, 5], index = ["a", "b", "c", "d", "e"]),
        "Four" : pd.Series([1, 2, 3, 4, 5, 6], index = ["a", "b", "c", "d", "e", "f"])}
df = pd.DataFrame(data)
print(df)
print(20* "#")
print(df.iloc[3])
```

```python
# Seleção por Slice
print(df[0:2])
```

```python
df = df[0:5]
df
```
