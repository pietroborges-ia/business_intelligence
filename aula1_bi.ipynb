```python
# --- ATIVIDADE 1: CONVERSOR DE ESCALA DE DADOS ---
# Volume de dados acumulado em 2020 segundo a Aula 1 (64.2 Trilhões de GB)
gigabytes_2020 = 64200000000000
# Conversões computacionais (Base 1024 / Padrão da Ciência da Computação)
megabytes = gigabytes_2020 * 1024
terabytes = gigabytes_2020 / 1024
petabytes = terabytes / 1024
exabytes = petabytes / 1024
zettabytes = exabytes / 1024
print("=" * 55)
print(" RELATÓRIO DE DIMENSIONAMENTO DE DADOS GLOBAIS")
print("=" * 55)
print(f"Volume Base em GB: {gigabytes_2020:,.0f} GB")
print(f"Volume em Megabytes (MB): {megabytes:,.0f} MB")
print(f"Volume em Terabytes (TB): {terabytes:,.2f} TB")
print(f"Volume em Petabytes (PB): {petabytes:,.2f} PB")
print(f"Volume em Zettabytes (ZB): {zettabytes:.2f} ZB")
print("=" * 55)
```
