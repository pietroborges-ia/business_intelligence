```python
import pandas as pd
salario = pd.read_csv("Nome,Idade,Salário.csv")

salario["Salário"] = (
    salario["Salário"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)


print(salario)
```

```python
mais_novo = salario.loc[salario["Idade"] == salario ["Idade"].min()]
print(mais_novo)
```
