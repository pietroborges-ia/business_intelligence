#### Recebe o nome e o ano de nascimento. E imprime "Olá {fulano}, vi que você tem {X} anos!"

```python
def funcao(nome, ano):
    idade = 2026 - ano
    print(f"Olá {nome}, sua idade é: {idade}")

nome = input("Digite seu nome: ")
ano = int(input("Digite o ano de nascimento: "))

funcao(nome, ano)
```
